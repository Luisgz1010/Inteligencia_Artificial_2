Problema de las gradientes:
En redes profundas, el gradiente puede volverse demasiado pequeño o demasiado grande. Con funciones como la sigmoide aparece la saturación, haciendo más difícil que la red aprenda.
Inicialización:
Para evitar estos problemas se usan métodos como Glorot, He y LeCun, que ayudan a comenzar los pesos con valores adecuados.
Funciones de activación:
Se utilizan funciones como ELU, SELU, GELU y Swish para mantener mejores señales durante el entrenamiento y facilitar el aprendizaje de la red.
Normalización y estabilización:
Se busca que los datos y activaciones tengan valores más controlados, normalmente centrando, escalando y desplazando los valores.
Aprendizaje por transferencia:
Se aprovecha una red que ya fue entrenada. Se pueden congelar las capas iniciales y modificar solamente las capas superiores para adaptarla a una nueva tarea.
Preentrenamiento y regularización:
Primero se puede entrenar una red con muchos datos y después adaptarla a una tarea específica. Esto permite aprovechar mejor el conocimiento aprendido y la regularización ayuda a reducir el sobreajuste.
