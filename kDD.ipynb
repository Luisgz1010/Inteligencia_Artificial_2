Resumen Sesion De Caos a Conocimineto KDD


DEL CAOS AL CONOCIMIENTO: KDD
1. El valor de los datos

Actualmente vivimos una explosión de datos, debido al crecimiento de Internet, las redes sociales, las empresas, los dispositivos y los sistemas digitales. Sin embargo, tener grandes cantidades de datos no significa tener conocimiento.

Los datos crudos son información que todavía no ha sido organizada, limpiada o analizada. Por esta razón, necesitan ser procesados de forma sistemática para poder encontrar información útil, identificar patrones y tomar mejores decisiones.

El objetivo de KDD es transformar los datos en conocimiento útil y estructurado.

Datos → Información → Conocimiento → Toma de decisiones

2. Origen y conceptos del KDD

KDD significa Knowledge Discovery in Databases, que en español se traduce como Descubrimiento de Conocimiento en Bases de Datos.

El concepto fue evolucionando con el tiempo. Entre los principales referentes se encuentran:

1989  Piatetsky-Shapiro: contribuyó al desarrollo y definición del concepto de descubrimiento de conocimiento y minería de datos.
1996 Fayyad: formalizó y explicó ampliamente el proceso de KDD, mostrando que la minería de datos es solamente una etapa dentro de un proceso más grande.
2016 Grady: representa una etapa posterior en la evolución y aplicación de estos conceptos en el contexto moderno de los datos.

Una idea fundamental es que KDD es todo el proceso o la "fábrica" completa, mientras que la minería de datos es solamente una parte de esa fábrica.

Por eso, hacer minería de datos sin realizar correctamente las etapas anteriores puede producir resultados incorrectos o poco confiables.

3. Proceso de KDD

El proceso KDD permite transformar datos crudos en conocimiento estructurado. Se realiza mediante una serie de etapas que permiten refinar los datos progresivamente.

Las cinco grandes etapas son:

1. Selección

Se identifican y seleccionan los datos que realmente son necesarios para el análisis.

En esta etapa se busca aislar la información útil y descartar variables o datos que no aportan al objetivo.

2. Preprocesamiento o reprocesamiento

Los datos seleccionados pueden contener errores, valores faltantes, duplicados o información inconsistente.

Por eso se realiza una limpieza y corrección de los datos para mejorar su calidad.

3. Transformación

Los datos se organizan o modifican para que puedan ser analizados correctamente.

Se pueden cambiar formatos, crear nuevas variables o realizar diferentes operaciones para preparar los datos para la minería.

4. Minería de datos

Es una de las etapas más importantes del proceso. Aquí se aplican métodos matemáticos, estadísticos y computacionales para encontrar patrones y relaciones en los datos.

Algunos modelos utilizados son:

Clasificación: permite asignar datos a diferentes categorías.
Regresión: permite predecir valores numéricos.
K-means: permite agrupar datos que tienen características similares.

La minería de datos busca descubrir patrones que no son evidentes a simple vista.

5. Evaluación

Finalmente, se analizan los patrones y resultados obtenidos para determinar si son válidos, útiles, relevantes y comprensibles.

No todos los patrones encontrados son importantes. Por eso es necesario evaluar los resultados antes de convertirlos en conocimiento y utilizarlos para tomar decisiones.

Resumen del proceso

Selección → Preprocesamiento → Transformación → Minería de datos → Evaluación → Conocimiento

En otras palabras:

Datos crudos → Datos limpios → Datos preparados → Patrones → Evaluación → Conocimiento útil

4. Aplicaciones reales: KDD en acción

KDD tiene aplicaciones en muchos sectores de la sociedad y las empresas.

Finanzas

Se utiliza para la detección de fraudes, analizando patrones de transacciones y comportamientos que pueden indicar operaciones sospechosas.

Salud

Puede utilizarse para apoyar el diagnóstico temprano, identificando patrones en grandes cantidades de información médica.

Retail o comercio

Se utiliza para el análisis de la canasta de compra. Esto permite descubrir qué productos suelen ser comprados juntos y ayuda a las empresas a mejorar sus estrategias de ventas.

5. El futuro del conocimiento

El desarrollo de KDD y de la minería de datos también genera nuevos desafíos relacionados con la ética, la seguridad y la privacidad.

Los algoritmos pueden analizar grandes cantidades de información y descubrir patrones sobre las personas. Esto puede ser útil, pero también puede representar un riesgo si los datos se utilizan de manera irresponsable.

Por esta razón, es importante:

Proteger la información de los usuarios.
Mantener la seguridad de los datos.
Respetar la privacidad.
Utilizar los algoritmos de manera ética y responsable.
Evitar que el reconocimiento de patrones sea utilizado para invadir la privacidad de las personas.

La ética y los algoritmos son cada vez más importantes porque el conocimiento obtenido de los datos puede influir en decisiones que afectan directamente a las personas.

CONCLUSIÓN

El KDD es un proceso que permite transformar grandes cantidades de datos crudos en conocimiento útil para la toma de decisiones. Para lograrlo, se seleccionan, limpian, transforman y analizan los datos, y posteriormente se evalúan los resultados.

La minería de datos es solamente una etapa del KDD, aunque es una de las más importantes porque permite descubrir patrones mediante modelos matemáticos y computacionales.

Finalmente, el crecimiento del KDD debe estar acompañado de responsabilidad, ya que el uso de datos también implica retos relacionados con la ética, la seguridad y la privacidad.
