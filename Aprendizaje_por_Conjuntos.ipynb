# APRENDIZAJE POR CONJUNTOS

El aprendizaje por conjuntos (*Ensemble Learning*) es una técnica de Machine Learning que combina varios modelos para obtener una predicción más confiable. La idea es que varios modelos, aunque individualmente puedan cometer errores, al trabajar juntos pueden producir un resultado más fuerte.

## 1. Introducción a conjuntos

Un conjunto está formado por varios modelos de Machine Learning que trabajan juntos. Estos modelos pueden ser iguales o diferentes. La combinación de sus resultados permite aprovechar las fortalezas de cada modelo y reducir algunos errores.

La idea principal es:

**Varios aprendices débiles → combinación → aprendiz fuerte.**

## 2. Clasificadores de votación

Los clasificadores de votación combinan las predicciones de varios modelos para obtener una predicción global unificada.

### Votación dura (Hard Voting)

Cada modelo realiza una predicción y se cuenta cuál categoría obtiene más votos. La categoría con mayor cantidad de votos se convierte en la predicción final.

### Votación blanda (Soft Voting)

Cada modelo proporciona probabilidades para las diferentes categorías. Las probabilidades se combinan y se selecciona la categoría con mayor probabilidad.

**Hard Voting:** cuenta votos.  
**Soft Voting:** combina probabilidades.

## 3. Bagging, Pasting y OOB

**Bagging** significa *Bootstrap Aggregating*. Consiste en entrenar varios modelos utilizando diferentes muestras de los datos. Las muestras se crean **con reemplazo**, por lo que algunos datos pueden aparecer varias veces.

**Pasting** es similar al Bagging, pero las muestras se seleccionan **sin reemplazo**.

**OOB (Out-of-Bag)** son los datos que no fueron seleccionados para entrenar un determinado modelo mediante Bagging. Estos datos pueden utilizarse para evaluar el rendimiento del modelo.

**Bagging → con reemplazo**  
**Pasting → sin reemplazo**  
**OOB → datos que quedaron fuera**

## 4. Bosques aleatorios

Un **Random Forest o Bosque Aleatorio** es un conjunto de muchos árboles de decisión. Cada árbol realiza una predicción y posteriormente se combinan los resultados de todos los árboles.

Por ejemplo, si la mayoría de los árboles predice "Fraude", el bosque puede tomar "Fraude" como resultado final.

Random Forest utiliza diferentes muestras de los datos y características aleatorias para crear árboles variados y obtener una predicción más robusta.

## 5. Algoritmos de Boosting

El **Boosting** construye modelos de manera secuencial. Cada nuevo modelo intenta corregir o mejorar los errores cometidos por los modelos anteriores.

El proceso puede representarse así:

**Modelo 1 → identifica errores → Modelo 2 → corrige errores → Modelo 3 → mejora el resultado.**

Algunos algoritmos conocidos son:

- AdaBoost
- Gradient Boosting
- XGBoost
- LightGBM

La principal diferencia con Bagging es que en Bagging los modelos se entrenan principalmente de forma independiente, mientras que en Boosting los modelos se construyen de manera progresiva para mejorar los resultados anteriores.

## 6. Stacking y Metaaprendizaje

**Stacking** combina diferentes modelos y utiliza un modelo adicional llamado **meta-modelo** para aprender cómo combinar sus predicciones.

Por ejemplo:

**Árbol de decisión + SVM + Regresión → Meta-modelo → Predicción final**

El meta-modelo analiza las predicciones de los modelos anteriores y aprende cuál combinación puede producir un mejor resultado.

El **metaaprendizaje** busca que un sistema aprenda sobre el propio proceso de aprendizaje, identificando qué modelos o estrategias pueden funcionar mejor dependiendo del problema.

## Conclusión

El aprendizaje por conjuntos busca mejorar las predicciones combinando varios modelos. Las principales técnicas son la votación, Bagging, Pasting, Random Forest, Boosting y Stacking.

En resumen:

**Voting → combina votos.**  
**Bagging → utiliza muestras con reemplazo.**  
**Pasting → utiliza muestras sin reemplazo.**  
**OOB → utiliza datos que quedaron fuera para evaluar.**  
**Random Forest → combina muchos árboles.**  
**Boosting → corrige errores progresivamente.**  
**Stacking → utiliza un meta-modelo para combinar otros modelos.**

La idea fundamental es que **la combinación de varios modelos puede producir un sistema más robusto y confiable que utilizar un único modelo.**
