# REDUCCIÓN DE DIMENSIONALIDAD

La **reducción de dimensionalidad** es una técnica de Machine Learning que busca disminuir la cantidad de características o variables de un conjunto de datos, intentando conservar la información más importante.

Por ejemplo, si tenemos un conjunto de datos con 100 características, podemos intentar reducirlo a 2 o 3 dimensiones para facilitar su análisis, visualización y procesamiento.

## 1. La maldición de la dimensionalidad

La **maldición de la dimensionalidad** ocurre cuando trabajamos con una cantidad muy grande de características.

A medida que aumenta el número de dimensiones, los datos se vuelven más dispersos en el espacio. Esto puede hacer que los modelos tengan más dificultades para encontrar patrones y aumente el riesgo de **sobreajuste (overfitting)**.

En espacios de alta dimensión:

- Los datos están más dispersos.
- Los modelos pueden necesitar más datos.
- Aumenta el costo computacional.
- Puede aumentar el riesgo de sobreajuste.
- La visualización de los datos se vuelve más complicada.

Por eso, reducir las dimensiones puede ayudar a simplificar los datos y facilitar el aprendizaje.

## 2. Reduciendo el espacio

Una forma de reducir la dimensionalidad es realizar una **proyección**.

La idea es proyectar datos que se encuentran en un espacio de muchas dimensiones hacia un espacio con menos dimensiones.

Por ejemplo:

**3 dimensiones → 2 dimensiones**

o

**100 dimensiones → 2 o 3 dimensiones**

Esto permite visualizar los datos y trabajar con ellos de una manera más sencilla.

### Manifold Learning

El **Manifold Learning** es un conjunto de técnicas que buscan encontrar estructuras o patrones dentro de datos que tienen muchas dimensiones.

La idea es que aunque los datos estén representados en muchas dimensiones, pueden encontrarse sobre una estructura de menor dimensión llamada **manifold**.

Por ejemplo, un conjunto de puntos puede parecer complejo en un espacio tridimensional, pero realmente estar organizado siguiendo una superficie de menor dimensión.

## 3. PCA y datos masivos

**PCA (Principal Component Analysis)** es una de las técnicas más conocidas para reducción de dimensionalidad.

PCA busca encontrar nuevas variables llamadas **componentes principales**, que permiten conservar la mayor cantidad posible de información o variación de los datos utilizando menos dimensiones.

Cuando trabajamos con conjuntos de datos muy grandes, utilizar PCA sobre todos los datos puede requerir bastante memoria y tiempo.

Una alternativa es utilizar métodos que trabajen con **pequeños minibatches o mini lotes de datos**.

Una técnica relacionada es **Incremental PCA (IPCA)**, que permite procesar los datos por partes en lugar de cargar todo el conjunto de datos en memoria al mismo tiempo.

Esto resulta útil para conjuntos de datos grandes.

## 4. LLE y otras técnicas

### LLE – Locally Linear Embedding

**LLE (Locally Linear Embedding)** es una técnica de reducción de dimensionalidad que busca conservar las relaciones entre los puntos que están cerca unos de otros.

La idea principal es que cada punto puede representarse aproximadamente utilizando sus vecinos más cercanos.

LLE intenta mantener esas relaciones locales cuando los datos se transforman a un espacio de menor dimensión.

### MDS – Multidimensional Scaling

**MDS** busca representar los datos en un espacio de menor dimensión intentando conservar las **distancias entre los puntos**.

Por ejemplo, si dos puntos estaban cerca en el espacio original, MDS intenta mantenerlos cerca después de la reducción.

### Isomap

**Isomap** también busca reducir la dimensionalidad, pero intenta conservar las **distancias geodésicas**, es decir, las distancias que siguen la estructura o superficie donde se encuentran los datos.

Isomap puede ser útil cuando los datos tienen una estructura curva o no lineal.

## Comparación

| Técnica | Idea principal |
|---|---|
| **PCA** | Conserva la mayor cantidad posible de variación |
| **Incremental PCA** | Aplica PCA por partes o mini lotes |
| **LLE** | Conserva relaciones entre vecinos cercanos |
| **MDS** | Conserva las distancias entre puntos |
| **Isomap** | Conserva distancias geodésicas de la estructura |

## Conclusión

La reducción de dimensionalidad permite trabajar con conjuntos de datos que tienen muchas características de una manera más sencilla y eficiente.

La **maldición de la dimensionalidad** muestra que trabajar con demasiadas dimensiones puede aumentar la dispersión de los datos, el costo computacional y el riesgo de sobreajuste.

Técnicas como **PCA, Incremental PCA, LLE, MDS e Isomap** permiten reducir el número de dimensiones intentando conservar la información importante de los datos.

En resumen:

**Muchas dimensiones → datos dispersos y complejos → reducción dimensional → datos más simples de analizar.**
