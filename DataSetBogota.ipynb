```python
import kagglehub
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt

# Descargar dataset de viviendas en Bogotá
path = kagglehub.dataset_download(
    "juandavsnchez/real-estatehousing-colombia-bogota"
)

print("Dataset descargado correctamente.")
print("Ubicación del dataset:")
print(path)
```

```python
print("Archivos encontrados:")

for archivo in os.listdir(path):
    print("-", archivo)
```

```python
# Buscar archivos CSV
archivos_csv = [
    archivo for archivo in os.listdir(path)
    if archivo.lower().endswith(".csv")
]

print("Archivos CSV encontrados:")
print(archivos_csv)

# Seleccionar el primer archivo CSV
archivo_csv = os.path.join(path, archivos_csv[0])

# Cargar datos
df = pd.read_csv(archivo_csv)

print("\nDataset cargado correctamente.")
print("Filas:", df.shape[0])
print("Columnas:", df.shape[1])
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("Columnas después de la limpieza:")

for columna in df.columns:
    print("-", columna)
```

```python
print("Valores faltantes por columna:\n")

print(df.isnull().sum())
```

```python
print("Cantidad de registros antes:", len(df))

df = df.drop_duplicates()

print("Cantidad de registros después:", len(df))
```

```python
posibles_area = [
    "area",
    "area_m2",
    "area_m2_construidos",
    "superficie"
]

posibles_habitaciones = [
    "habitaciones",
    "habitacion",
    "rooms",
    "room",
    "bedrooms"
]

posibles_banos = [
    "banos",
    "baños",
    "bathrooms",
    "bathroom",
    "baths"
]

posibles_precio = [
    "precio",
    "price",
    "valor",
    "value",
    "precio_venta"
]


def buscar_columna(posibles, columnas):
    for nombre in posibles:
        if nombre in columnas:
            return nombre
    return None


col_area = buscar_columna(posibles_area, df.columns)
col_habitaciones = buscar_columna(posibles_habitaciones, df.columns)
col_banos = buscar_columna(posibles_banos, df.columns)
col_precio = buscar_columna(posibles_precio, df.columns)


print("Área:", col_area)
print("Habitaciones:", col_habitaciones)
print("Baños:", col_banos)
print("Precio:", col_precio)
```

```python
if col_area is None:
    print("No se encontró automáticamente la columna de área.")

if col_habitaciones is None:
    print("No se encontró automáticamente la columna de habitaciones.")

if col_banos is None:
    print("No se encontró automáticamente la columna de baños.")

if col_precio is None:
    print("No se encontró automáticamente la columna de precio.")
```

```python
if col_area is None:
    col_area = 'área' # Manually assign the correct column name after previous cell failed to identify it

datos = df[
    [
        col_area,
        col_habitaciones,
        col_banos,
        col_precio
    ]
].copy()

datos.columns = [
    "area",
    "habitaciones",
    "banos",
    "precio"
]

print("Dataset de viviendas creado correctamente.")

datos.head()
```

```python
datos["area"] = pd.to_numeric(
    datos["area"],
    errors="coerce"
)

datos["habitaciones"] = pd.to_numeric(
    datos["habitaciones"],
    errors="coerce"
)

datos["banos"] = pd.to_numeric(
    datos["banos"],
    errors="coerce"
)

datos["precio"] = pd.to_numeric(
    datos["precio"],
    errors="coerce"
)

datos.info()
```

```python
print("Viviendas antes de eliminar datos vacíos:", len(datos))

datos = datos.dropna()

print("Viviendas después de eliminar datos vacíos:", len(datos))
```

```python
datos = datos[
    (datos["area"] > 0) &
    (datos["habitaciones"] > 0) &
    (datos["banos"] >= 0) &
    (datos["precio"] > 0)
]

print("Cantidad final de viviendas:", len(datos))
```

```python
X = datos[
    [
        "area",
        "habitaciones",
        "banos"
    ]
]

y = datos["precio"]

print("Variables independientes:")
print(X.head())

print("\nVariable objetivo:")
print(y.head())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Viviendas para entrenamiento:", len(X_train))
print("Viviendas para prueba:", len(X_test))
```

```python
from sklearn.linear_model import LinearRegression

modelo_lineal = LinearRegression()

modelo_lineal.fit(
    X_train,
    y_train
)

print("Modelo de regresión lineal entrenado correctamente.")
```

```python
print("Intercepto:")
print(modelo_lineal.intercept_)

print("\nCoeficientes:")

for nombre, coeficiente in zip(
    X.columns,
    modelo_lineal.coef_
):
    print(nombre, ":", coeficiente)
```

```python
predicciones_lineal = modelo_lineal.predict(
    X_test
)

print("Primeras predicciones:\n")

for i in range(10):
    print(
        "Precio real:",
        y_test.iloc[i],
        "| Precio predicho:",
        predicciones_lineal[i]
    )
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

mae_lineal = mean_absolute_error(
    y_test,
    predicciones_lineal
)

mse_lineal = mean_squared_error(
    y_test,
    predicciones_lineal
)

rmse_lineal = mse_lineal ** 0.5

r2_lineal = r2_score(
    y_test,
    predicciones_lineal
)

print("===== REGRESIÓN LINEAL =====")

print("MAE :", mae_lineal)
print("MSE :", mse_lineal)
print("RMSE:", rmse_lineal)
print("R²  :", r2_lineal)
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    y_test,
    predicciones_lineal,
    alpha=0.5
)

plt.xlabel("Precio real")
plt.ylabel("Precio predicho")

plt.title(
    "Regresión lineal: precio real vs. precio predicho"
)

plt.show()
```

```python
from sklearn.preprocessing import PolynomialFeatures

poly = PolynomialFeatures(
    degree=2,
    include_bias=False
)

X_train_poly = poly.fit_transform(
    X_train
)

X_test_poly = poly.transform(
    X_test
)

print(
    "Variables originales:",
    X_train.shape[1]
)

print(
    "Variables después de la transformación:",
    X_train_poly.shape[1]
)
```

```python
modelo_polinomial = LinearRegression()

modelo_polinomial.fit(
    X_train_poly,
    y_train
)

print(
    "Modelo de regresión polinomial entrenado correctamente."
)
```

```python
predicciones_poly = modelo_polinomial.predict(
    X_test_poly
)

print("Primeras predicciones:\n")

for i in range(10):
    print(
        "Precio real:",
        y_test.iloc[i],
        "| Precio predicho:",
        predicciones_poly[i]
    )
```

```python
mae_poly = mean_absolute_error(
    y_test,
    predicciones_poly
)

mse_poly = mean_squared_error(
    y_test,
    predicciones_poly
)

rmse_poly = mse_poly ** 0.5

r2_poly = r2_score(
    y_test,
    predicciones_poly
)

print("===== REGRESIÓN POLINOMIAL =====")

print("MAE :", mae_poly)
print("MSE :", mse_poly)
print("RMSE:", rmse_poly)
print("R²  :", r2_poly)
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    y_test,
    predicciones_poly,
    alpha=0.5
)

plt.xlabel("Precio real")
plt.ylabel("Precio predicho")

plt.title(
    "Regresión polinomial: precio real vs. precio predicho"
)

plt.show()
```

```python
resultados = pd.DataFrame({
    "Modelo": [
        "Regresión Lineal",
        "Regresión Polinomial"
    ],
    "MAE": [
        mae_lineal,
        mae_poly
    ],
    "MSE": [
        mse_lineal,
        mse_poly
    ],
    "RMSE": [
        rmse_lineal,
        rmse_poly
    ],
    "R2": [
        r2_lineal,
        r2_poly
    ]
})

resultados
```

```python
resultados_redondeados = resultados.copy()

resultados_redondeados[
    ["MAE", "MSE", "RMSE", "R2"]
] = resultados_redondeados[
    ["MAE", "MSE", "RMSE", "R2"]
].round(2)

resultados_redondeados
```

```python
nueva_vivienda_poly = poly.transform(
    nueva_vivienda
)

precio_poly = modelo_polinomial.predict(
    nueva_vivienda_poly
)

print("Características de la vivienda:")
print("Área: 100 m²")
print("Habitaciones: 3")
print("Baños: 2")

print("\nPrecio estimado:")

print(
    f"${precio_poly[0]:,.0f} COP"
)
```

```python
grados = [1, 2, 3]

resultados_grados = []

for grado in grados:

    poly_temp = PolynomialFeatures(
        degree=grado,
        include_bias=False
    )

    X_train_temp = poly_temp.fit_transform(
        X_train
    )

    X_test_temp = poly_temp.transform(
        X_test
    )

    modelo_temp = LinearRegression()

    modelo_temp.fit(
        X_train_temp,
        y_train
    )

    pred_temp = modelo_temp.predict(
        X_test_temp
    )

    mae_temp = mean_absolute_error(
        y_test,
        pred_temp
    )

    rmse_temp = mean_squared_error(
        y_test,
        pred_temp
    ) ** 0.5

    r2_temp = r2_score(
        y_test,
        pred_temp
    )

    resultados_grados.append({
        "Grado": grado,
        "MAE": mae_temp,
        "RMSE": rmse_temp,
        "R2": r2_temp
    })


resultados_grados = pd.DataFrame(
    resultados_grados
)

resultados_grados
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    resultados_grados["Grado"],
    resultados_grados["R2"],
    marker="o"
)

plt.xlabel("Grado del polinomio")
plt.ylabel("R²")

plt.title(
    "R² según el grado del polinomio"
)

plt.grid()

plt.show()
```
