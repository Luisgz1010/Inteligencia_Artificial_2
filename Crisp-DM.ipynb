CRISP-DM
1. Orígenes y conceptos

CRISP-DM significa Cross-Industry Standard Process for Data Mining, es decir, Proceso Estándar Interindustrial para la Minería de Datos.

CRISP-DM fue concebido a mediados de los años 90 por un consorcio formado por empresas como DaimlerChrysler, SPSS y NCR. En 1999 se publicó oficialmente como una metodología para organizar proyectos de minería de datos.

Su objetivo es proporcionar un proceso estructurado para desarrollar proyectos de análisis y minería de datos, desde la comprensión del problema del negocio hasta la implementación de los resultados.

CRISP-DM se caracteriza por ser un proceso iterativo, lo que significa que no siempre se sigue de manera completamente lineal. Si durante una fase se encuentra un problema, se puede regresar a una fase anterior para corregirlo.

2. Comprensión del negocio: los cimientos del éxito

La primera fase se concentra en entender qué necesita el negocio y qué problema se quiere solucionar.

Fase 1: Comprensión del negocio

En esta etapa se establecen:

Objetivos del negocio: qué quiere conseguir la organización.
Situación actual: qué está ocurriendo y qué información se tiene disponible.
Metas: resultados concretos que se quieren alcanzar.
Plan: cómo se utilizarán los datos para alcanzar los objetivos.

Por ejemplo, una empresa podría utilizar datos para detectar fraudes y evitar pérdidas de millones de pesos.

La idea principal es que el proyecto de datos debe comenzar con una necesidad real del negocio, y no simplemente con la intención de analizar datos.

3. Comprensión de los datos

Después de establecer el objetivo del negocio, es necesario conocer los datos disponibles.

Esta fase incluye:

Recolectar: obtener los datos de las diferentes fuentes.
Describir: conocer qué contienen los datos.
Explorar: buscar patrones, relaciones y características importantes.
Evaluar la calidad: identificar errores, datos faltantes, duplicados o inconsistencias.

Esta fase permite determinar si los datos disponibles realmente pueden ayudar a resolver el problema planteado.

4. Preparación de los datos

La preparación de datos suele ser una de las fases que requiere más trabajo.

Aquí se toma la información cruda y se transforma en un conjunto de datos integrado, limpio y preparado para el modelado.

Se pueden realizar actividades como:

Limpiar errores.
Eliminar duplicados.
Tratar valores faltantes.
Seleccionar variables importantes.
Unir información de diferentes fuentes.
Transformar los formatos de los datos.

Una idea fundamental es:

“Basura entra, basura sale” (Garbage In, Garbage Out).

Esto significa que si los datos utilizados son incorrectos o de mala calidad, los resultados del modelo también pueden ser incorrectos.

5. Modelado y evaluación
Fase 4: Modelado

Cuando los datos ya están preparados, comienza la parte más matemática y técnica.

Se seleccionan las técnicas de minería de datos o aprendizaje automático más adecuadas y se construyen modelos.

El proceso normalmente consiste en:

Elegir técnica → Construir modelo → Probar → Evaluar → Mejorar

Dependiendo del problema, se pueden utilizar modelos de clasificación, regresión, agrupamiento, predicción, entre otros.

Fase 5: Evaluación

Una vez construidos los modelos, se revisan los resultados para determinar si realmente cumplen los objetivos planteados al comienzo.

Se evalúan tanto:

Los resultados del modelo: ¿funciona correctamente?
El proceso: ¿se siguieron los pasos adecuados?
Los objetivos del negocio: ¿el modelo realmente ayuda a resolver el problema?

Si los resultados no son satisfactorios, se puede regresar a fases anteriores, por ejemplo, mejorar los datos o cambiar el modelo.

Resultados → Evaluación → Decisión → Próximos pasos

6. Despliegue y futuro
Fase 6: Despliegue

El objetivo final es llevar el modelo o los resultados a un entorno donde puedan utilizarse realmente.

El despliegue puede incluir:

Integración operativa: incorporar el modelo a los procesos de la empresa.
Monitoreo: controlar que el modelo siga funcionando correctamente.
Informes: comunicar los resultados.
Revisión: comprobar periódicamente si el modelo sigue siendo útil.

Un modelo no termina cuando se construye. Es necesario vigilarlo y actualizarlo porque los datos y las condiciones del negocio pueden cambiar.

7. CRISP-DM y la inteligencia artificial moderna

Con la evolución de la inteligencia artificial, CRISP-DM continúa siendo útil porque proporciona una estructura para organizar los proyectos de datos.

La IA puede automatizar algunas tareas, como la preparación de datos, la selección de modelos o el entrenamiento, pero siguen siendo necesarios:

Comprender el problema del negocio.
Tener datos de calidad.
Evaluar correctamente los resultados.
Supervisar los modelos.
Considerar aspectos éticos y de seguridad.

Por lo tanto, la IA puede hacer que algunas etapas sean más rápidas y automatizadas, pero no elimina la necesidad de un proceso organizado.

CONCLUSIÓN

CRISP-DM es una metodología que organiza los proyectos de minería de datos desde el problema del negocio hasta la implementación de una solución.

Sus principales fases son:

1. Comprensión del negocio → 2. Comprensión de los datos → 3. Preparación de los datos → 4. Modelado → 5. Evaluación → 6. Despliegue

Su principal ventaja es que permite trabajar de forma estructurada e iterativa, regresando a fases anteriores cuando sea necesario.

La idea central es:

Problema de negocio → Datos → Preparación → Modelo → Evaluación → Solución en producción
