Resumen de Inteligencia Artificial 2 Sesion 1
1. Hacia una IA avanzada
La Inteligencia Artificial busca crear modelos capaces de aprender a partir de grandes cantidades de datos.
Para desarrollar estos modelos se utiliza un ecosistema tecnológico
Python: lenguaje de programación utilizado para desarrollar modelos de IA.
Scikit-learn: biblioteca de Python que permite crear, entrenar y evaluar modelos de aprendizaje automático.
2. Preparación de datos
Antes de entrenar un modelo, es necesario preparar correctamente los datos. Una mala preparación puede provocar errores o incluso resultados catastróficos en el modelo.
Algunas técnicas importantes son:
Imputación: completar o reemplazar datos faltantes.
Escalamiento: colocar los datos en escalas comparables para que una variable no tenga demasiada influencia sobre las demás.
Reducción: disminuir la cantidad de variables o información innecesaria.
Clasificación: asignar los datos a diferentes categorías.
La preparación de datos es fundamental porque un modelo no puede ser mejor que la calidad de los datos con los que aprende.
3. Modelos ensamblados supervisados
Los modelos ensamblados combinan varios modelos de aprendizaje para obtener mejores resultados.

Entre los principales métodos están:

Kernel Trick: permite trabajar con problemas que no son linealmente separables transformando los datos a un espacio donde sea más fácil encontrar patrones.
Bagging: utiliza varios modelos, generalmente árboles de decisión, trabajando en paralelo. Su objetivo principal es reducir la varianza y obtener un modelo más estable.
Boosting: construye modelos de manera secuencial, donde cada nuevo modelo intenta corregir los errores de los anteriores.
4. Aprendizaje no supervisado
En el aprendizaje no supervisado, el modelo trabaja con datos que no tienen una respuesta o categoría previamente indicada.

Su objetivo es descubrir patrones ocultos, grupos o estructuras dentro de los datos.

Aquí también es importante la varianza, que indica cuánto pueden cambiar los resultados de un modelo dependiendo de los datos utilizados.

Además, es necesario identificar los datos críticos o ruido estadístico, ya que información irrelevante o demasiado diferente puede afectar negativamente los resultados.

5. Evaluación de modelos
Después de crear un modelo, debemos comprobar qué tan bien funciona. Para esto se estudian conceptos como:

Sesgo: ocurre cuando el modelo es demasiado simple y no logra representar correctamente los patrones de los datos.
Varianza: ocurre cuando el modelo se adapta demasiado a los datos de entrenamiento y tiene dificultades con datos nuevos.
El objetivo es encontrar un equilibrio entre sesgo y varianza para conseguir un modelo que generalice bien.

6. Búsqueda de los mejores parámetros
GridSearchCV prueba diferentes combinaciones de parámetros para encontrar la configuración que produce mejores resultados. Es una búsqueda exhaustiva, por lo que puede ser costosa y lenta cuando existen muchas combinaciones.

RandomizedSearchCV selecciona combinaciones de parámetros de forma aleatoria. Esto permite explorar el espacio de búsqueda utilizando menos recursos y puede ser mucho más rápido.

Métricas avanzadas
Para evaluar modelos de clasificación existen diferentes métricas.

La matriz de confusión permite comparar las predicciones del modelo con los resultados reales. A partir de ella podemos obtener:

Precisión : de todas las predicciones positivas realizadas por el modelo, cuántas fueron realmente positivas.
Exhaustividad: de todos los casos que realmente eran positivos, cuántos logró encontrar el modelo.
Estas métricas son importantes porque la exactitud por sí sola no siempre es suficiente para saber si un modelo funciona correctamente.

La idea principal de todo el tema es que para construir una buena Inteligencia Artificial no basta con elegir un algoritmo: hay que preparar bien los datos, escoger el modelo adecuado, controlar el sesgo y la varianza y finalmente evaluar sus resultados con métricas apropiadas.
